```python
import urllib.request
import shutil
import os 
import zipfile
import io
```

```python
#Latest Version of Synthea
#sample files containing 100 patient records in csv formats

url = "https://synthetichealth.github.io/synthea-sample-data/downloads/latest/synthea_sample_data_csv_latest.zip"

response = urllib.request.urlopen(url)

dir_path ="/Volumes/syntheamedalliondatapipeline/00_landing/data_sources/synthe/"

os.makedirs(dir_path, exist_ok=True)

with urllib.request.urlopen(url) as response:
    with zipfile.ZipFile(io.BytesIO(response.read())) as zip_ref:
        zip_ref.extractall(dir_path)

print("Data loading is finished...!")
```
